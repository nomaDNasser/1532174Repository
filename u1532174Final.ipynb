```python
# Damon Nasser
# u1532174


def main():
  Display_Title()
  Find_Climber()


import time
import csv

def print_slow(text) :
  for char in text:
    print(char, end='', flush=True)
    time.sleep(0.1)







def Display_Title():
  print_slow("""
  Welcome to the International Climbing Database""")





def Find_Climber():
  open("Big_One.csv", "r")
  Title = input(print_slow("Enter a Climber's First and Last Name"))
  with open("Big_One.csv", "r") as file:
    reader = csv.reader(file)
    for row in file:
      if Title in row:
        print_slow("Climber found:")
        print("".join(row))
        promptII()
      elif Title not in row:
        print_slow("Climber not found")
        main()

def promptII()
  print_slow("Please Enter the following information")
  First_Name = input(print_slow("Enter the Climber's First Name"))
  Last_Name = input(print_slow("Enter the Climber's Last Name"))
  Athlete_ID = input(print_slow("Enter the athlete ID. You can find this in the first row of the data we've outputed already."))
  print_slow("https://www.ifsc-climbing.org/athlete/"Athlete_ID"/"First_Name"-"Last_Name)
  print_slow("https://ifsc.results.info/athlete/"Athlete_ID)
  print_slow("https://boulderingmaster.com/player/"First_Name"-"Last_Name"/")
  print_slow("https://www.sportclimbingstats.com/climber/"First_Name"-"Last_Name"/")
  print_slow("Copy any of these links into your browser to find their respective pages on you athlete")



def Exit():
  print_slow("Exiting Library")
  exit()




if __name__ == '__main__':
  main()
```
